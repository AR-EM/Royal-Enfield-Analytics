# Case Study: Royal Enfield Motors India
## Data-Driven Transformation in Mid-Weight Motorcycle Retail & Operations

---

## Executive Summary

**Royal Enfield** (a unit of Eicher Motors Limited) is the global leader in the mid-size motorcycle segment (250cc–750cc). Over the past decade, Royal Enfield transformed from a niche heritage brand into a global motorcycle powerhouse, driven by modern platform architectures (J-series 350cc, 452cc Liquid-Cooled Sherpa, and 650cc Parallel Twins).

However, managing rapid retail expansion across 200 dealerships, 5,000 active customer segments, 20,000 vehicle transactions, 8,000 financed loans, 15,000 service visits, and 12,000 Genuine Motorcycle Accessory (GMA) transactions presents significant operational complexity.

This case study examines Royal Enfield's enterprise resource planning (ERP) analytics to address four critical business challenges:
1. **Sales & Finance Portfolio Optimization**: Balancing high-volume 350cc models with premium 650cc margin drivers.
2. **Aftermarket & Accessory Monetization**: Driving high-margin GMA penetration across customer segments.
3. **Service Quality & Warranty Cost Containment**: Reducing warranty failure rates on new engine platforms.
4. **CRM Lead Conversion Efficiency**: Optimizing dealer test ride fulfillment to drive retail sales.

---

## 1. Company Background & Market Position

Founded in 1901 in Redditch, UK, and manufacturing in Chennai, India since 1955, Royal Enfield is the world's oldest motorcycle brand in continuous production. 

### The Product Portfolio (15 Models across 5 Categories)
* **Modern Classics (350cc & 500cc)**: Bullet 350, Classic 350, Hunter 350, Classic 500, Bullet 500.
* **Roadsters & Cruisers**: Meteor 350, Super Meteor 650, Shotgun 650, Thunderbird 350X.
* **Adventure & Dual-Sport**: Himalayan 450, Guerrilla 450, Scram 411.
* **Cafe Racers**: Continental GT 650, Interceptor 650, Continental GT 535.

## 2. Key Commercial Challenges & Analytical Questions

### Challenge A: Dealer Network Performance & Sales Penetration
With 200 dealerships spanning Flagship Showrooms, Authorized Dealerships, and Studio Stores across 5 regions (North, South, East, West, Central), performance variance is stark.
* *Question for Analysts*: Which regions and dealership formats achieve the highest monthly target completion rate? Are Studio Stores in Tier 2/3 cities outperforming Flagship Showrooms on ROI?

### Challenge B: Vehicle Financing & Credit Risk
40% of all vehicle purchases are financed through institutional banking partners (HDFC, ICICI, SBI, Bajaj Finance, TVS Credit). 
* *Question for Analysts*: What is the average EMI burden relative to customer annual income across different age groups? Are higher interest rates impacting financing approval for first-time buyers?

### Challenge C: After-Sales Retention & Service Revenue
After-sales service generates steady recurring revenue and drives brand loyalty. However, servicing costs vary significantly between single-cylinder 350cc bikes and twin-cylinder 650cc bikes.
* *Question for Analysts*: Which motorcycle models exhibit the highest post-warranty service retention? What is the ratio of parts cost to labor cost across regular maintenance vs. breakdown repairs?

### Challenge D: Warranty Claim Rate & Component Reliability
Warranty claims account for substantial operational expenditure. Analyzing component failure categories (Engine/Transmission, Electrical, EFI, Suspension) is vital for quality control.
* *Question for Analysts*: Which engine platform demonstrates the highest warranty claim rate per 1,000 units sold? What is the average financial payout approval rate across component categories?

### Challenge E: CRM Lead Conversion (Test Rides)
Test rides are the primary conversion driver in motorcycle retail.
* *Question for Analysts*: What is the overall test-ride-to-sale conversion rate? Does a higher test ride feedback rating (4-5 stars) directly correlate with higher sales conversion speed?

# Royal Enfield ERP Analytics - Pandas Data Manipulation & Reshaping Guide

This document is a comprehensive technical guide for students and analysts working with Python's **Pandas** library on the Royal Enfield ERP dataset package. 

It covers fundamental and advanced operations: combining datasets (`pd.merge()`, `df.join()`), reshaping data (`pd.pivot_table()`, `pd.melt()`, `unstack()`), and applying these techniques to answer the core business challenges in `royal_enfiled.ipynb`.

---

## 1. Core Data Combination Techniques

### 1.1 `pd.merge()` vs `df.join()`
* **`pd.merge()`**: Flexible function used to combine DataFrames based on common key columns (e.g. `Customer_ID`, `Dealer_ID`, `Bike_ID`). Supports `how='inner'|'left'|'right'|'outer'`.
* **`df.join()`**: Method used to join DataFrames on their Index (or key column to index).


---

## 2. Reshaping & Restructuring Data

### 2.1 `pd.pivot_table()`
Pivot tables aggregate tabular data across 2 dimensions (index rows and header columns).
* `index`: Column(s) to group by on the vertical axis (rows).
* `columns`: Column(s) to group by on the horizontal axis (headers).
* `values`: Column(s) to aggregate.
* `aggfunc`: Aggregation function (`'sum'`, `'mean'`, `'count'`, `'median'`, etc.).
* `margins=True`: Includes grand totals across rows and columns.

### 2.2 `pd.melt()` (Unpivoting)
`pd.melt()` transforms wide-format DataFrames into long-format DataFrames (unpivoting).
* `id_vars`: Column(s) to keep fixed as identifiers.
* `value_vars`: Column(s) to unpivot into rows.
* `var_name`: Name of the new column for variable labels.
* `value_name`: Name of the new column for values.

#### Example: Unpivoting Service Costs
```python
service = pd.read_csv('Service.csv')

# Unpivot Parts_Cost and Labor_Cost into long format
service_costs_long = pd.melt(
    service, 
    id_vars=['Service_ID', 'Service_Type', 'Bike_ID'], 
    value_vars=['Parts_Cost', 'Labor_Cost'],
    var_name='Cost_Component', 
    value_name='Amount_INR'
)
```



| Pandas Function | Purpose in Royal Enfield ERP Analytics |
| :--- | :--- |
| `pd.merge()` | Inner and left-joining transactional tables (`Sales`, `Service`, `Finance`, `TestRide`) with master lookup tables (`Customers`, `Dealers`, `Motorcycles`). |
| `pd.pivot_table()` | Multi-dimensional aggregation calculating Target Completion %, EMI-to-Income %, Conversion Rate %, and Claim Rate per 1,000 units. |
| `pd.melt()` | Unpivoting `Parts_Cost` and `Labor_Cost` from wide columns into a single `Cost_Component` column for comparative ratio analysis. |
| `unstack()` / `stack()` | Converting multi-level index series into clean, readable report tables. |
